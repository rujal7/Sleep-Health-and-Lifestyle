## Sleep Health and Lifestyle Data Analysis

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Load Dataset and Dataset Understanding

```python
df=pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")
df
```

```python
print("=== Dataset Information ===")
df.info()

print("\n=== Statistical Summary ===")
print(df.describe())

print("\n=== Missing Values in Each Column ===")
print(df.isnull().sum())
```

## 3. BMI Category vs Sleep Disorder

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="BMI Category", hue="Sleep Disorder")
plt.title("BMI Category vs Sleep Disorder")
plt.show()
```

## 4. Correlation Heatmap

```python
numeric = df.select_dtypes(include="number")

plt.figure(figsize=(10,8))
sns.heatmap(numeric.corr(), annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## 5. Average Quality of sleep by sleep Disoder

```python
quality = df.groupby("Sleep Disorder")["Quality of Sleep"].mean().reset_index()

plt.figure(figsize=(6,4))

sns.barplot(
    data=quality,
    x="Sleep Disorder",
    y="Quality of Sleep"
)

plt.title("Average Quality of Sleep by Sleep Disorder")
plt.ylabel("Average Quality Score")

plt.show()
```

## 6. Average Daily Steps by Sleep Disoder

```python
quality = df.groupby("Sleep Disorder")["Daily Steps"].mean().reset_index()

plt.figure(figsize=(6,4))

sns.barplot(
    data=quality,
    x="Sleep Disorder",
    y="Daily Steps",
    palette="viridis"
)

plt.title("Average Daily Steps by Sleep Disorder")
plt.ylabel("Average Daily Steps")

plt.show()
```

## 7. Quality of Sleep and Sleep Disorder

```python
plt.figure(figsize=(15,9))
sns.barplot(data=df, x='Occupation', y='Quality of Sleep', hue='Sleep Disorder')
plt.show()
```

## 8. Sleep Disorder by Occupation

```python
plt.figure(figsize=(15,9))
sns.countplot(
    data=df,
    x="Occupation",
    palette="Set2",
    hue="Sleep Disorder"
)
plt.title('Sleep Disorder by Occupation')
plt.show()
```

## 9. Gender vs Sleep Disorder

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Gender",
    hue="Sleep Disorder",
    palette="Pastel1"
)

plt.title("Gender vs Sleep Disorder")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

## 10. Average Stress Level by Sleep Disorder

```python
stress = df.groupby("Sleep Disorder")["Stress Level"].mean().reset_index()

plt.figure(figsize=(6,4))

sns.barplot(
    data=stress,
    x="Sleep Disorder",
    y="Stress Level",
    palette="rocket"
)

plt.title("Average Stress Level by Sleep Disorder")
plt.ylabel("Average Stress Level")

plt.show()
```

## 11. Physical Activity vs Sleep Disorder

```python
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x="Sleep Disorder", y="Physical Activity Level", palette="magma")
plt.title("Physical Activity vs Sleep Disorder")
plt.show()
```

## 12. Stress Level vs Sleep Duration

```python
plt.figure(figsize=(7,5))
sns.scatterplot(
    data=df,
    x="Stress Level",
    y="Sleep Duration",
    hue="Sleep Disorder"
)
plt.title("Stress Level vs Sleep Duration")
plt.show()
```

## 13. Sleep Disorder and Age

```python
sns.boxplot(data=df,x="Sleep Disorder",y="Age", palette="coolwarm")
plt.show()
```

## 14. Distibution of Sleep Disorders

```python
sleep_disorder = df["Sleep Disorder"].fillna("None")
sleep_count = sleep_disorder.value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    sleep_count,
    labels=sleep_count.index,
    autopct='%1.1f%%',
    
)

plt.title("Distribution of Sleep Disorders")
plt.show()
```
